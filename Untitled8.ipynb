```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

ages = np.random.normal(28, 18, 1000)

ages = np.clip(ages, 0, 100)
```

```python
plt.figure(figsize=(10,6))

plt.hist(ages,
         bins=20,
         color="skyblue",
         edgecolor="black")

plt.title("Age Distribution of Population")

plt.xlabel("Age")

plt.ylabel("Number of People")

plt.grid(alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt

age_groups = ["0-20", "21-64", "65+"]

population = [512,807,98]

plt.figure(figsize=(8,5))

plt.bar(age_groups,
        population,
        color=["gold","royalblue","deeppink"])

plt.title("India Population by Age Group")

plt.xlabel("Age Groups")

plt.ylabel("Population (Millions)")

plt.show()
```
